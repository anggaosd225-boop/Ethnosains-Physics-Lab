```python
#@title ▶ 1. Setup (pustaka & tema visual) { display-mode: "form" }
# Hanya memakai pustaka yang sudah tersedia di Google Colab: numpy, matplotlib, ipywidgets.
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as w
from IPython.display import display, clear_output, HTML
from functools import lru_cache
from matplotlib.patches import Circle, Rectangle, FancyArrowPatch, Wedge, Polygon, Arc
import traceback

# Palet warna bertema tanah liat (terakota) Kasongan
C = dict(bg='#FBF3E6', paper='#FFFAF0', ink='#3B2417', clay='#B5532D', clay2='#D98E5F',
         sand='#E8D5B5', teal='#1F7A8C', gold='#E0A030', red='#C0392B', grey='#8C7B6B',
         green='#3E8E41')

plt.rcParams.update({
    'figure.facecolor': C['bg'], 'axes.facecolor': C['paper'], 'axes.edgecolor': C['ink'],
    'axes.labelcolor': C['ink'], 'text.color': C['ink'], 'xtick.color': C['ink'],
    'ytick.color': C['ink'], 'font.size': 10, 'axes.grid': True, 'grid.alpha': 0.25,
    'axes.spines.top': False, 'axes.spines.right': False, 'figure.dpi': 90,
    'axes.titleweight': 'bold', 'axes.titlesize': 11,
})

G = 9.81  # percepatan gravitasi (m/s²)
print("Setup selesai ✓")
```

```python
#@title ▶ 2. Inti Fisika (fungsi murni tanpa UI) { display-mode: "form" }
# Semua perhitungan fisika ada di sini, terpisah dari antarmuka, sehingga dapat diuji sendiri.
# Satuan SI: m, kg, s, N, N·m, rad. Konvensi tanda: berlawanan arah jarum jam (CCW) = POSITIF.

# ---------------- A. Torsi ----------------
def torque(r, F, theta_deg):
    """τ = r F sin θ ; θ = sudut dari vektor posisi r ke vektor gaya F (CCW positif)."""
    return r * F * np.sin(np.radians(theta_deg))

def torque_from_vectors(r_vec, F_vec):
    """Komponen-z dari τ = r × F (definisi vektor, dipakai sebagai pembanding)."""
    return r_vec[0] * F_vec[1] - r_vec[1] * F_vec[0]

# ---------------- B. Momen inersia ----------------
def I_disk(M, R):        return 0.5 * M * R**2      # cakram pejal, sumbu melalui pusat
def I_ring(M, R):        return M * R**2            # cincin tipis, sumbu melalui pusat
def I_rod_center(M, L):  return M * L**2 / 12.0     # batang homogen, sumbu di tengah
def I_rod_end(M, L):     return M * L**2 / 3.0      # batang homogen, sumbu di ujung
def I_parallel(I_cm, M, d): return I_cm + M * d**2  # teorema sumbu sejajar

def clay_inertia(m, r, shape):
    """Tanah liat simetris terhadap sumbu: 'disk' = gumpalan (cakram pejal), 'ring' = dinding tipis."""
    return I_disk(m, r) if shape == 'disk' else I_ring(m, r)

def wheel_inertia(M_w, R_w, m_c, r_c, shape):
    """Momen inersia total roda putar (cakram) + tanah liat di atasnya (aditif)."""
    return I_disk(M_w, R_w) + clay_inertia(m_c, r_c, shape)

def I_numeric(shape, M, size, n=600):
    """Pembanding independen: I = Σ mᵢrᵢ² dari diskretisasi benda (bukan dari rumus tabel)."""
    if shape == 'disk':
        xs = np.linspace(-size, size, n); X, Y = np.meshgrid(xs, xs)
        r2 = (X**2 + Y**2)[X**2 + Y**2 <= size**2]
        return M * np.mean(r2)
    if shape == 'ring':
        return M * size**2                       # semua massa berada pada jarak R
    if shape == 'rod_center':                    # panjang = size
        x = (np.arange(n) + 0.5) / n * size - size / 2
        return M * np.mean(x**2)
    if shape == 'rod_end':
        x = (np.arange(n) + 0.5) / n * size
        return M * np.mean(x**2)
    raise ValueError(shape)

# ---------------- C. Dinamika rotasi (roda putar) ----------------
def _net_torque(omega, tau_app, tau_f, b):
    """Στ pada roda: torsi dorong − gesekan Coulomb (τ_f, melawan gerak) − redaman viskos (bω)."""
    if omega > 1e-9:  return tau_app - tau_f - b * omega
    if omega < -1e-9: return tau_app + tau_f - b * omega
    if abs(tau_app) <= tau_f: return 0.0          # gesekan statis menahan roda
    return tau_app - np.sign(tau_app) * tau_f

@lru_cache(maxsize=512)
def simulate_wheel(I, tau_push, t_push, tau_f, b, T, dt=0.002):
    """Integrasi numerik (RK4) dari  I dω/dt = Στ,  dθ/dt = ω.
    Torsi dorong konstan selama 0 ≤ t < t_push, lalu 0. Mengembalikan dict hasil + kerja/energi."""
    n = int(round(T / dt))
    t = np.arange(n + 1) * dt
    om = np.zeros(n + 1); th = np.zeros(n + 1); al = np.zeros(n + 1)
    W_app = 0.0; W_fric = 0.0
    for k in range(n):
        ta = tau_push if (t[k] + 0.5 * dt) < t_push else 0.0
        w0 = om[k]
        k1 = _net_torque(w0, ta, tau_f, b) / I
        al[k] = k1
        if abs(w0) < 2 * dt * tau_f / I:             # dekat ω=0: langkah Euler + logika "berhenti"
            w1 = w0 + dt * k1
        else:                                        # RK4 biasa
            k2 = _net_torque(w0 + 0.5 * dt * k1, ta, tau_f, b) / I
            k3 = _net_torque(w0 + 0.5 * dt * k2, ta, tau_f, b) / I
            k4 = _net_torque(w0 + dt * k3, ta, tau_f, b) / I
            w1 = w0 + dt * (k1 + 2 * k2 + 2 * k3 + k4) / 6.0
        if abs(ta) <= tau_f and (w0 * w1 < 0 or abs(w1) < 1e-9):   # mencapai/melewati nol tanpa dorongan → berhenti
            w1 = 0.0
        om[k + 1] = w1
        dth = 0.5 * dt * (w0 + w1)
        th[k + 1] = th[k] + dth
        W_app += ta * dth
        W_fric += tau_f * abs(dth) + b * 0.5 * (w0 * w0 + w1 * w1) * dt
    al[n] = al[n - 1]
    return dict(t=t, omega=om, theta=th, alpha=al, dt=dt, I=I, tau_push=tau_push, t_push=t_push,
                tau_f=tau_f, b=b, W_app=W_app, W_fric=W_fric, K_end=0.5 * I * om[-1]**2)

def wheel_summary(sim):
    om, t = sim['omega'], sim['t']
    om_max = float(om.max())
    t_stop = None
    if om_max > 0:
        idx = np.where((t >= sim['t_push']) & (om <= 0))[0]
        if len(idx): t_stop = float(t[idx[0]])
    return dict(alpha0=float(sim['alpha'][0]), omega_max=om_max, t_stop=t_stop,
                turns=float(sim['theta'][-1] / (2 * np.pi)))

def measure_alpha(I, tau):
    """α yang DIUKUR dari simulasi (kemiringan grafik ω–t tanpa gesekan), bukan dari rumus τ/I."""
    s = simulate_wheel(float(round(I, 6)), float(round(tau, 6)), 1.0, 0.0, 0.0, 1.0)
    return float(np.polyfit(s['t'], s['omega'], 1)[0])

# ---------------- D. Kesetimbangan benda tegar ----------------
def beam_reactions(L, Mb, xa, xb, loads):
    """Balok homogen (panjang L, massa Mb) ditumpu di xa dan xb; loads = [(m, x), ...].
    ΣFy=0 dan Στ=0 terhadap titik A  ⇒  N_A, N_B (newton)."""
    d = xb - xa
    tau_w = Mb * G * (L / 2 - xa) + sum(m * G * (x - xa) for m, x in loads)
    NB = tau_w / d
    NA = Mb * G + sum(m * G for m, _ in loads) - NB
    return NA, NB

def net_torque_about(p, forces):
    """Στ terhadap titik p; forces = [(x, Fy), ...] (gaya vertikal). τ_z = (x−p)·Fy, CCW positif."""
    return sum((x - p) * Fy for x, Fy in forces)

def beam_forces(L, Mb, loads, supports=None):
    f = [(L / 2, -Mb * G)] + [(x, -m * G) for m, x in loads]
    if supports: f += supports
    return f

def tipping_info(L, Mb, xa, xb, loads, NA, NB):
    """Jika salah satu tumpuan tak mampu menahan (N<0), balok mulai berotasi terhadap tumpuan lain:
    kembalikan (titik putar, torsi total, I terhadap titik itu, α awal)."""
    if NA < 0: p = xb
    elif NB < 0: p = xa
    else: return None
    tau = net_torque_about(p, beam_forces(L, Mb, loads))
    I = I_rod_center(Mb, L) + Mb * (L / 2 - p)**2 + sum(m * (x - p)**2 for m, x in loads)
    return dict(pivot=p, tau=tau, I=I, alpha0=tau / I)
```

```python
#@title ▶ 3. Uji Mandiri Fisika (dijalankan otomatis) { display-mode: "form" }
# Setiap uji membandingkan model dengan hasil analitik / metode independen. Hasil NYATA dari eksekusi.

def run_self_tests():
    res = []
    def add(name, ok, detail): res.append((name, bool(ok), detail))
    rng = np.random.default_rng(42)

    # T1: τ = rF sinθ  ≡  komponen-z dari r × F
    err = 0
    for _ in range(200):
        r = rng.uniform(0.05, 1.5); F = rng.uniform(0, 100); th = rng.uniform(-180, 180); phi = rng.uniform(0, 360)
        rv = r * np.array([np.cos(np.radians(phi)), np.sin(np.radians(phi))])
        Fv = F * np.array([np.cos(np.radians(phi + th)), np.sin(np.radians(phi + th))])
        err = max(err, abs(torque(r, F, th) - torque_from_vectors(rv, Fv)))
    add("Torsi: rF sinθ = komponen-z r × F (200 kasus acak)", err < 1e-9, f"galat maks = {err:.1e} N·m")

    # T2: kasus khusus torsi
    ok = abs(torque(0.5, 10, 0)) < 1e-12 and abs(torque(0.5, 10, 180)) < 1e-9 and abs(torque(0.5, 10, 90) - 5) < 1e-12 \
         and torque(0.5, 10, -90) < 0
    add("Torsi: nol di θ=0°/180°, maksimum di 90°, tanda negatif di −90°", ok, "τ(90°)=rF=5,0 N·m")

    # T3: teorema sumbu sejajar
    M, L = 3.0, 1.7
    d = abs(I_parallel(I_rod_center(M, L), M, L / 2) - I_rod_end(M, L))
    add("Momen inersia: I_ujung = I_tengah + M(L/2)² (sumbu sejajar)", d < 1e-12, f"selisih = {d:.1e}")

    # T4: rumus tabel vs penjumlahan Σmr² numerik
    cases = [('disk', I_disk(2, .5), 2, .5), ('ring', I_ring(2, .5), 2, .5),
             ('rod_center', I_rod_center(2, 1.0), 2, 1.0), ('rod_end', I_rod_end(2, 1.0), 2, 1.0)]
    worst = max(abs(I_numeric(s, M_, sz) - Ia) / Ia for s, Ia, M_, sz in cases)
    add("Momen inersia: 4 rumus = Σmᵢrᵢ² dari diskretisasi benda", worst < 0.01, f"galat relatif maks = {worst*100:.3f} %")

    # T5: dinamika — ω(t) simulasi vs solusi analitik (b=0)
    I = 0.25; tau = 1.2; tp = 2.0; tf = 0.3
    s = simulate_wheel(I, tau, tp, tf, 0.0, 15.0)
    a = (tau - tf) / I; wp = a * tp; t_stop = tp + wp * I / tf
    def w_exact(t):
        return np.where(t < tp, a * t, np.maximum(wp - (tf / I) * (t - tp), 0.0))
    e = np.max(np.abs(s['omega'] - w_exact(s['t'])))
    add("Roda putar: ω(t) simulasi = solusi analitik (dorong + gesekan konstan)", e < 5e-3, f"selisih maks = {e:.1e} rad/s")
    sm = wheel_summary(s)
    add("Roda putar: waktu berhenti simulasi = analitik", sm['t_stop'] is not None and abs(sm['t_stop'] - t_stop) < 0.01,
        f"simulasi {sm['t_stop']:.3f} s vs analitik {t_stop:.3f} s")

    # T6: Στ = Iα  →  α terukur dari kemiringan ω–t
    worst = 0
    for I_ in (0.1, 0.5, 1.3):
        for tau_ in (0.5, 2.0, 7.0):
            worst = max(worst, abs(measure_alpha(I_, tau_) - tau_ / I_) / (tau_ / I_))
    add("Στ = Iα: α terukur dari kemiringan ω–t = τ/I (9 kombinasi)", worst < 1e-6, f"galat relatif maks = {worst:.1e}")

    # T7: kerja–energi
    s = simulate_wheel(0.35, 2.0, 3.0, 0.2, 0.05, 12.0)
    lhs = s['W_app'] - s['W_fric']; rhs = s['K_end']
    scale = max(s['W_app'], 1e-9)
    add("Kerja–energi: W_dorong − W_gesek = ΔE_k rotasi", abs(lhs - rhs) / scale < 2e-3,
        f"W_dorong={s['W_app']:.3f} J, W_gesek={s['W_fric']:.3f} J, E_k akhir={rhs:.4f} J")

    # T8: gesekan statis menahan roda bila τ_dorong ≤ τ_gesek
    s = simulate_wheel(0.3, 0.25, 2.0, 0.3, 0.0, 5.0)
    add("Gesekan statis: roda tetap diam bila τ_dorong ≤ τ_gesek", float(np.abs(s['omega']).max()) == 0.0, "ω ≡ 0")

    # T9: I lebih besar → α lebih kecil (torsi sama)
    Is = [wheel_inertia(8, .25, 2, .08, 'disk'), wheel_inertia(8, .25, 2, .22, 'ring')]
    al = [measure_alpha(i, 2.0) for i in Is]
    add("Sebaran massa: tanah liat melebar (cincin) → I naik, α turun", Is[1] > Is[0] and al[1] < al[0],
        f"I: {Is[0]:.4f}→{Is[1]:.4f} kg·m²; α: {al[0]:.2f}→{al[1]:.2f} rad/s²")

    # T10: kesetimbangan — ΣFy = 0 dan Στ = 0 terhadap TITIK MANA PUN
    cnt = 0; worst = 0
    while cnt < 100:
        L = rng.uniform(1, 3); xa, xb = np.sort(rng.uniform(0, L, 2))
        if xb - xa < 0.2: continue
        loads = [(rng.uniform(0, 15), rng.uniform(0, L)) for _ in range(2)]; Mb = rng.uniform(1, 10)
        NA, NB = beam_reactions(L, Mb, xa, xb, loads)
        if NA < 0 or NB < 0: continue
        f = beam_forces(L, Mb, loads, [(xa, NA), (xb, NB)])
        r = [abs(sum(Fy for _, Fy in f))] + [abs(net_torque_about(p, f)) for p in (0, L, xa, xb, rng.uniform(-2, 5))]
        worst = max(worst, max(r)); cnt += 1
    add("Kesetimbangan: ΣFy=0 & Στ=0 terhadap 5 titik acuan (100 konfigurasi acak)", worst < 1e-9, f"residu maks = {worst:.1e}")

    # T11: kasus simetris & batas guling
    NA, NB = beam_reactions(2.0, 4.0, 0.3, 1.7, [(5, 0.7), (5, 1.3)])
    add("Kesetimbangan: konfigurasi simetris → N_A = N_B = W/2", abs(NA - NB) < 1e-9 and abs(NA + NB - 14 * G) < 1e-9,
        f"N_A=N_B={NA:.2f} N")
    m_crit = 4.0 * (1.7 - 1.0) / (2.0 - 1.7)
    NAc, _ = beam_reactions(2.0, 4.0, 0.3, 1.7, [(m_crit, 2.0)])
    NAs, _ = beam_reactions(2.0, 4.0, 0.3, 1.7, [(m_crit * 0.95, 2.0)])
    NAt, NBt = beam_reactions(2.0, 4.0, 0.3, 1.7, [(m_crit * 1.05, 2.0)])
    add("Batas guling: N_A = 0 tepat di m = M(x_B−L/2)/(x−x_B)", abs(NAc) < 1e-9 and NAs > 0 and NAt < 0,
        f"m_kritis = {m_crit:.3f} kg")
    return res

TEST_RESULTS = run_self_tests()
for name, ok, detail in TEST_RESULTS:
    print(("✅ " if ok else "❌ ") + name + "  →  " + detail)
print(f"\nRingkasan: {sum(o for _, o, _ in TEST_RESULTS)}/{len(TEST_RESULTS)} uji lulus")
```

```python
#@title ▶ 4. Pembantu Antarmuka (Panel, slider, kartu HTML) { display-mode: "form" }

def show(fig):
    """Tampilkan figur ke Output aktif lalu tutup (mencegah kebocoran memori)."""
    plt.show()
    plt.close(fig)

def S(desc, v, lo, hi, step, fmt='.2f'):
    return w.FloatSlider(value=v, min=lo, max=hi, step=step, description=desc, readout_format=fmt,
                         continuous_update=False, style={'description_width': '170px'},
                         layout=w.Layout(width='430px'))

def header(text, color=None):
    return w.HTML(f"<div style='font-weight:700;color:{color or C['clay']};font-family:sans-serif;"
                  f"font-size:14px;margin:6px 0 2px 0'>{text}</div>")

def card(title, rows, color=None):
    color = color or C['clay']
    trs = ''.join(f"<tr><td style='padding:2px 14px 2px 0;color:#6b5646'>{a}</td>"
                  f"<td style='padding:2px 0;font-weight:600'>{b}</td></tr>" for a, b in rows)
    return (f"<div style='flex:1;min-width:300px;background:#FFF8EC;border-left:5px solid {color};padding:8px 14px;"
            f"margin:6px 8px 6px 0;border-radius:6px;color:#3B2417;font-family:sans-serif;font-size:13.5px'>"
            f"<b>{title}</b><table style='margin-top:4px;color:#3B2417'>{trs}</table></div>")

def cards(*items):
    display(HTML("<div style='display:flex;flex-wrap:wrap'>" + ''.join(items) + "</div>"))

def sense(tau, eps=1e-9):
    if abs(tau) < eps: return "tidak berotasi (Στ = 0)"
    return "berlawanan arah jarum jam (+)" if tau > 0 else "searah jarum jam (−)"

def info_box(title, html_body):
    acc = w.Accordion(children=[w.HTML(f"<div style='font-family:sans-serif;font-size:14px;line-height:1.55;color:#3B2417;"
                                       f"background:#FFF8EC;padding:10px 14px;border-radius:6px'>{html_body}</div>")])
    acc.set_title(0, title)
    try: acc.selected_index = None
    except Exception: pass
    return acc

def lab_guide(body_html):
    """Kotak lipat 'Panduan eksperimen' (Prediksi → Amati → Jelaskan) di dasar tiap tab."""
    return info_box("🧪 Panduan eksperimen: Prediksi → Amati → Jelaskan (klik untuk membuka)", body_html)

def arrow(ax, p, v, color, lw=2.5, ls='-', ms=16, z=8):
    ax.add_patch(FancyArrowPatch((p[0], p[1]), (p[0] + v[0], p[1] + v[1]), arrowstyle='-|>', mutation_scale=ms,
                                 color=color, lw=lw, ls=ls, zorder=z, shrinkA=0, shrinkB=0, clip_on=False))

class Panel:
    """Pengikat kontrol → fungsi gambar. Gambar ulang hanya saat nilai berubah; galat ditampilkan, bukan menghentikan aplikasi."""
    def __init__(self, controls, draw_fn):
        self.controls, self.draw_fn = controls, draw_fn
        self.out = w.Output(); self._lock = False
        for c in controls.values():
            c.observe(self._changed, names='value')
    def _changed(self, _):
        if not self._lock: self.refresh()
    def values(self):
        return {k: c.value for k, c in self.controls.items()}
    def refresh(self):
        with self.out:
            clear_output(wait=True)
            try:
                self.draw_fn(**self.values())
            except Exception:
                print("⚠ Terjadi galat saat menggambar (laporkan teks berikut):"); traceback.print_exc()
    def set(self, **kw):
        self._lock = True
        try:
            for k, v in kw.items(): self.controls[k].value = v
        finally:
            self._lock = False
        self.refresh()
```

```python
#@title ▶ 5. Modul ① Momen Gaya (Torsi) { display-mode: "form" }

def draw_torque(F1, r1, th1, F2, r2, th2, phi):
    ph = np.radians(phi); u = np.array([np.cos(ph), np.sin(ph)]); nv = np.array([-u[1], u[0]])
    forces = [(F1, r1, th1, C['clay'], 'F₁', 'd₁'), (F2, r2, th2, C['teal'], 'F₂', 'd₂')]
    fig = plt.figure(figsize=(14, 5.6), constrained_layout=True)
    gs = fig.add_gridspec(1, 3, width_ratios=[1.8, 1, 0.8])
    ax = fig.add_subplot(gs[0]); ax.set_aspect('equal'); ax.axis('off')
    ax.set_xlim(-1.3, 1.3); ax.set_ylim(-1.3, 1.3)

    # tuas (batang) + poros
    p0, p1 = -0.12 * u, 1.05 * u
    ax.plot([p0[0], p1[0]], [p0[1], p1[1]], color=C['ink'], lw=17, solid_capstyle='round', zorder=1)
    ax.plot([p0[0], p1[0]], [p0[1], p1[1]], color=C['sand'], lw=14, solid_capstyle='round', zorder=2)
    for rr in (0.25, 0.5, 0.75, 1.0):
        q = rr * u
        ax.plot([q[0] - .035 * nv[0], q[0] + .035 * nv[0]], [q[1] - .035 * nv[1], q[1] + .035 * nv[1]], color=C['ink'], lw=1, zorder=3)
    ax.add_patch(Circle((0, 0), 0.055, fc=C['ink'], zorder=9))
    ax.text(0, -0.17, "poros", ha='center', fontsize=8, color=C['grey'])

    taus = []; arms = []
    for F, r, th, col, nm, dn in forces:
        tau = torque(r, F, th); taus.append(tau); arms.append(r * abs(np.sin(np.radians(th))))
        if F <= 0: continue
        P = r * u; ang = ph + np.radians(th); fh = np.array([np.cos(ang), np.sin(ang)])
        Lf = 0.1 + 0.6 * F / 100
        ax.plot([P[0]], [P[1]], 'o', color=col, ms=7, zorder=10)
        ax.plot([P[0], P[0] + 0.4 * u[0]], [P[1], P[1] + 0.4 * u[1]], color=C['grey'], lw=1, ls='--', zorder=4)  # acuan θ
        arrow(ax, P, Lf * fh, col, lw=3.2)
        lab = P + (Lf + 0.16) * fh
        ax.text(lab[0], lab[1], f"{nm} = {F:.0f} N", color=col, fontsize=10, fontweight='bold', ha='center', va='center')
        Fp = F * np.sin(np.radians(th))                         # komponen tegak lurus tuas
        if abs(Fp) > 0.5:
            arrow(ax, P, 0.6 * Fp / 100 * nv, col, lw=1.6, ls=':', ms=10, z=7)
        if abs(th) > 1:                                         # busur sudut θ
            a1, a2 = sorted([phi, phi + th])
            ax.add_patch(Arc(P, 0.42, 0.42, theta1=a1, theta2=a2, color=col, lw=1.6, zorder=6))
            mid = np.radians(phi + th / 2); tp_ = P + 0.36 * np.array([np.cos(mid), np.sin(mid)])
            ax.text(tp_[0], tp_[1], f"θ={th:.0f}°", color=col, fontsize=8.5, ha='center', va='center')
        if abs(np.sin(np.radians(th))) > 0.03:                  # lengan gaya d = r sinθ
            s = -np.dot(P, fh); Q = P + s * fh
            ax.plot([P[0], Q[0]], [P[1], Q[1]], color=col, lw=1, ls='--', alpha=.6, zorder=3)
            ax.plot([0, Q[0]], [0, Q[1]], color=col, lw=2.2, ls=':', zorder=5)
            ax.text(Q[0] * .5, Q[1] * .5, f" {dn}", color=col, fontsize=9, style='italic')

    tot = sum(taus)
    ax.set_title(f"Στ = {tot:+.2f} N·m  →  {sense(tot)}", color=(C['teal'] if tot > 0 else C['red'] if tot < 0 else C['grey']))

    # kurva τ(θ)
    ax2 = fig.add_subplot(gs[1]); th_arr = np.linspace(-180, 180, 361)
    ax2.plot(th_arr, torque(r1, F1, th_arr), color=C['clay'], lw=2, label='τ₁ (F₁, r₁)')
    ax2.plot([th1], [taus[0]], 'o', color=C['clay'], ms=9, zorder=5)
    if F2 > 0:
        ax2.plot(th_arr, torque(r2, F2, th_arr), color=C['teal'], lw=2, label='τ₂ (F₂, r₂)')
        ax2.plot([th2], [taus[1]], 'o', color=C['teal'], ms=9, zorder=5)
    ax2.axhline(0, color=C['ink'], lw=.8); ax2.set_xticks(range(-180, 181, 90))
    ax2.set_xlabel("θ  (derajat)"); ax2.set_ylabel("τ  (N·m)"); ax2.set_title("τ = r F sin θ  terhadap θ")
    ax2.legend(loc='lower right', fontsize=8)

    # batang torsi
    ax3 = fig.add_subplot(gs[2]); names = ['τ₁', 'τ₂', 'Στ']; vals = [taus[0], taus[1], tot]
    ax3.bar(names, vals, color=[C['teal'] if v >= 0 else C['red'] for v in vals], edgecolor=C['ink'])
    ax3.axhline(0, color=C['ink'], lw=.8); ax3.set_ylabel("N·m"); ax3.set_title("Penjumlahan torsi (dengan tanda)")
    for i, v in enumerate(vals):
        ax3.text(i, v, f"{v:+.1f}", ha='center', va='bottom' if v >= 0 else 'top', fontsize=9, fontweight='bold')
    show(fig)

    rows1 = [("Rumus", f"τ₁ = r·F·sinθ = {r1:.2f} × {F1:.0f} × sin({th1:.0f}°)"),
             ("Torsi", f"τ₁ = {taus[0]:+.2f} N·m"), ("Lengan gaya", f"d₁ = r·|sinθ| = {arms[0]:.2f} m"),
             ("Komponen ⟂ tuas", f"F₁·sinθ = {F1*np.sin(np.radians(th1)):+.1f} N")]
    rows2 = [("Rumus", f"τ₂ = {r2:.2f} × {F2:.0f} × sin({th2:.0f}°)"), ("Torsi", f"τ₂ = {taus[1]:+.2f} N·m"),
             ("Lengan gaya", f"d₂ = {arms[1]:.2f} m")]
    rows3 = [("Στ", f"{tot:+.2f} N·m"), ("Arah rotasi", sense(tot)),
             ("Catatan", "orientasi tuas (φ) tidak mengubah torsi")]
    cards(card("Gaya F₁", rows1, C['clay']), card("Gaya F₂", rows2, C['teal']), card("Hasil", rows3, C['ink']))

TORQUE_PRESETS = {
    "Awal: gaya tegak lurus tuas": dict(F1=40, r1=0.8, th1=90, F2=0, r2=0.4, th2=90, phi=20),
    "Gaya searah tuas (τ = 0)": dict(F1=40, r1=0.8, th1=0, F2=0, r2=0.4, th2=90, phi=20),
    "Lengan lebih panjang → torsi lebih besar": dict(F1=40, r1=1.0, th1=90, F2=0, r2=0.4, th2=90, phi=20),
    "Dua gaya berlawanan: Στ = 0": dict(F1=20, r1=0.8, th1=90, F2=40, r2=0.4, th2=-90, phi=20),
    "Gaya miring 30°": dict(F1=40, r1=0.8, th1=30, F2=0, r2=0.4, th2=90, phi=20),
}

def build_torque():
    c = dict(F1=S('Gaya F₁ (N)', 40, 0, 100, 1, '.0f'), r1=S('Jarak r₁ dari poros (m)', 0.8, 0.1, 1.0, 0.05),
             th1=S('Sudut θ₁ (derajat)', 90, -180, 180, 5, '.0f'),
             F2=S('Gaya F₂ (N)', 0, 0, 100, 1, '.0f'), r2=S('Jarak r₂ dari poros (m)', 0.4, 0.1, 1.0, 0.05),
             th2=S('Sudut θ₂ (derajat)', 90, -180, 180, 5, '.0f'), phi=S('Orientasi tuas φ (derajat)', 20, 0, 360, 5, '.0f'))
    panel = Panel(c, draw_torque)
    dd = w.Dropdown(options=list(TORQUE_PRESETS), description='Skenario cepat:', style={'description_width': '120px'},
                    layout=w.Layout(width='520px'))
    dd.observe(lambda ch: panel.set(**TORQUE_PRESETS[ch['new']]), names='value')
    info = info_box("📘 Konsep & asumsi (klik untuk membuka)", """
<b>Torsi (momen gaya)</b> mengukur kecenderungan gaya membuat benda berotasi terhadap suatu poros:
<b>τ = r F sin θ</b>, dengan θ = sudut dari vektor posisi <i>r</i> (poros → titik tangkap gaya) ke vektor gaya <i>F</i>.
Setara: τ = F × d, dengan <i>d = r sin θ</i> adalah <b>lengan gaya</b> (jarak tegak lurus poros ke garis kerja gaya).<br>
<b>Konvensi tanda:</b> berlawanan arah jarum jam = positif; searah jarum jam = negatif.<br>
<b>Yang dapat dieksplorasi:</b> (1) gaya searah tuas (θ=0°) tidak menghasilkan torsi; (2) torsi maksimum saat θ=90°;
(3) memperbesar r atau F memperbesar τ; (4) dua gaya dapat saling meniadakan (Στ = 0).<br>
<b>Konteks:</b> tuas ini adalah model umum (mis. mendorong/memutar benda terhadap sebuah poros), bukan gambaran alat tertentu pengrajin.""")
    guide = lab_guide('''
<b>Langkah:</b>
<ol><li><b>Prediksi</b> (tulis di buku): pada sudut θ₁ berapa torsi τ₁ paling besar? Kapan nol? Apa yang terjadi bila θ₁ negatif?</li>
<li><b>Amati</b>: pilih skenario <i>Awal: gaya tegak lurus tuas</i>, lalu geser θ₁ ke 0°, 30°, 90°, −90°. Lihat titik di grafik tengah, batang torsi (biru kehijauan = positif, merah = negatif), dan kartu <i>Gaya F₁</i>.</li>
<li><b>Jelaskan</b>: hubungkan dengan <b>lengan gaya</b> d₁ = r·|sin θ| (garis titik-titik dari poros). Mengapa lengan gaya nol saat θ = 0°?</li></ol>
<b>Tantangan:</b>
<ol><li>Dengan F₁ = 40 N dan r₁ = 0,8 m, berapa τ₁ saat θ₁ = 30°? Cek di kartu (seharusnya 16 N·m).</li>
<li>Atur F₁ = 40 N, r₁ = 0,8 m, θ₁ = 90° (τ₁ = 32 N·m). Pasang r₂ = 0,4 m dan θ₂ = −90°. Berapa F₂ agar Στ = 0? (cek: 80 N)</li>
<li>Putar <i>orientasi tuas φ</i>. Apakah torsi berubah? Mengapa?</li></ol>
<b>Membaca hasil:</b> tanda + = berlawanan arah jarum jam, − = searah jarum jam. Judul diagram menunjukkan Στ dan arah rotasinya.''')
    ui = w.VBox([header("① Momen Gaya (Torsi): τ = r F sin θ"),
                 w.HTML("<div style='font-family:sans-serif;font-size:13.5px;color:#3B2417'>🎯 <b>Tugas:</b> Prediksikan dulu, lalu ubah sudut θ₁. "
                        "Pada sudut berapa torsi paling besar? Kapan nol? Apa yang terjadi bila arah gaya dibalik?</div>"),
                 dd, w.HBox([w.VBox([c['F1'], c['r1'], c['th1']]), w.VBox([c['F2'], c['r2'], c['th2'], c['phi']])]),
                 panel.out, guide, info])
    panel.refresh()
    return ui, panel
```

```python
#@title ▶ 9. Modul ⑤ Kesetimbangan Benda Tegar { display-mode: "form" }

def draw_equilibrium(L, Mb, xa, xb, m1, x1, m2, x2):
    xa, xb, x1, x2 = [float(np.clip(v, 0, L)) for v in (xa, xb, x1, x2)]
    if xb - xa < 0.05:
        print("Jarak kedua tumpuan terlalu dekat: atur x_B > x_A."); return
    loads = [(m1, x1), (m2, x2)]
    NA, NB = beam_reactions(L, Mb, xa, xb, loads)
    ok = NA >= -1e-9 and NB >= -1e-9
    Wt = (Mb + m1 + m2) * G
    fl = beam_forces(L, Mb, loads, [(xa, NA), (xb, NB)] if ok else None)
    resF = sum(Fy for _, Fy in fl); resA = net_torque_about(xa, fl); resB = net_torque_about(xb, fl); resC = net_torque_about(L / 2, fl)
    tip = tipping_info(L, Mb, xa, xb, loads, NA, NB)

    fig = plt.figure(figsize=(13.5, 7.2), constrained_layout=True)
    gs = fig.add_gridspec(2, 1, height_ratios=[1.25, 1])
    ax = fig.add_subplot(gs[0]); ax.grid(False)
    for sp in ('left',): ax.spines[sp].set_visible(False)
    ax.set_yticks([]); ax.set_xlim(-0.3, L + 0.3); ax.set_ylim(-1.45, 1.7)
    Fmax = max(Wt, abs(NA), abs(NB), Mb * G, m1 * G, m2 * G, 1e-9); ks = 1.0 / Fmax
    # balok (miring sedikit bila tidak setimbang hanya sebagai penanda, bukan hasil simulasi)
    ax.add_patch(Rectangle((0, -0.07), L, 0.14, fc=C['sand'], ec=C['ink'], lw=2, zorder=3))
    for x in (xa, xb):
        ax.add_patch(Polygon([(x, -0.07), (x - 0.09, -0.45), (x + 0.09, -0.45)], fc=C['grey'], ec=C['ink'], alpha=.55, zorder=2))
    # berat balok
    arrow(ax, (L / 2, -0.07), (0, -Mb * G * ks), C['grey'], lw=3)
    ax.text(L / 2 + 0.03, -0.07 - Mb * G * ks * 0.5, f"W_balok = {Mb*G:.1f} N", ha='left', va='center', fontsize=9, color=C['grey'], fontweight='bold')
    ax.plot([L / 2], [0], 'x', color=C['ink'], ms=8, zorder=6)
    # beban (gerabah, ilustrasi)
    for i, (m, x) in enumerate(loads, 1):
        if m <= 0: continue
        wd = 0.07 + 0.012 * m
        ax.add_patch(Polygon([(x - wd * .7, .07), (x + wd * .7, .07), (x + wd, .07 + .22), (x - wd, .07 + .22)], fc=C['clay'], ec=C['ink'], alpha=.6, zorder=4))
        ytop = .07 + .22
        arrow(ax, (x, ytop + m * G * ks), (0, -m * G * ks), C['red'], lw=3)
        ax.text(x, ytop + m * G * ks + .04, f"m{i}g = {m*G:.1f} N\n(x{i} = {x:.2f} m)", ha='center', va='bottom', fontsize=9, color=C['red'], fontweight='bold')
    # reaksi tumpuan
    for nm, x, N in (("A", xa, NA), ("B", xb, NB)):
        if N >= -1e-9:
            arrow(ax, (x, -0.07 - N * ks), (0, N * ks), C['teal'], lw=3.5)
            ax.text(x, -0.07 - N * ks - 0.05, f"N_{nm} = {N:.1f} N", ha='center', va='top', fontsize=9.5, color=C['teal'], fontweight='bold')
            ax.text(x, -0.07 - N * ks - 0.27, f"tumpuan {nm} (x = {x:.2f} m)", ha='center', va='top', fontsize=8.5, color=C['grey'])
        else:
            ax.text(x, -0.55, f"N_{nm} = {N:.1f} N  < 0\n(tumpuan tak bisa menarik)", ha='center', va='top', fontsize=9.5, color=C['red'], fontweight='bold')
            ax.text(x, -0.95, f"tumpuan {nm} (x = {x:.2f} m)", ha='center', va='top', fontsize=8.5, color=C['grey'])
    ax.set_xlabel("posisi sepanjang balok  x (m)")
    if ok:
        ax.set_title("✓ SETIMBANG STATIS: ΣFy = 0 dan Στ = 0", color=C['green'])
    else:
        side = "A" if NA < 0 else "B"
        ax.set_title(f"✗ TIDAK SETIMBANG: tumpuan {side} lepas (N_{side} < 0) → balok berotasi terhadap tumpuan {'B' if side == 'A' else 'A'}", color=C['red'])

    ax2 = fig.add_subplot(gs[1]); xx = np.linspace(0, L, 240)
    NAx = np.array([beam_reactions(L, Mb, xa, xb, [(m1, x), (m2, x2)])[0] for x in xx])
    NBx = np.array([beam_reactions(L, Mb, xa, xb, [(m1, x), (m2, x2)])[1] for x in xx])
    ax2.fill_between(xx, min(NAx.min(), NBx.min(), 0), max(NAx.max(), NBx.max()), where=(NAx >= 0) & (NBx >= 0), color=C['green'], alpha=.12, label='daerah setimbang')
    ax2.plot(xx, NAx, color=C['teal'], lw=2, label='N_A'); ax2.plot(xx, NBx, color=C['clay'], lw=2, label='N_B')
    ax2.axhline(0, color=C['ink'], lw=1); ax2.axvline(x1, color=C['ink'], ls='--', lw=1)
    ax2.plot([x1, x1], [NA, NB], 'o', color=C['ink'], ms=8, zorder=6)
    ax2.set_xlabel("posisi beban 1, x₁ (m)  [massa lain tetap]"); ax2.set_ylabel("gaya tumpuan (N)")
    ax2.set_title("Seberapa jauh beban 1 boleh digeser sebelum balok terguling?"); ax2.legend(loc='best', fontsize=9)
    show(fig)

    rows = [("N_A", f"{NA:.2f} N"), ("N_B", f"{NB:.2f} N"), ("Berat total W", f"{Wt:.2f} N"),
            ("Cek ΣFy", f"{resF:+.1e} N" if ok else "(tumpuan lepas)")]
    chk = [("Στ terhadap A", f"{resA:+.1e} N·m"), ("Στ terhadap B", f"{resB:+.1e} N·m"), ("Στ terhadap pusat balok", f"{resC:+.1e} N·m"),
           ("Kesimpulan", "ΣFx = 0 (tak ada gaya horizontal), ΣFy = 0, Στ = 0 terhadap titik mana pun" if ok else "syarat kesetimbangan tidak terpenuhi")]
    items = [card("Gaya tumpuan", rows, C['teal']), card("Verifikasi syarat kesetimbangan" if ok else "Σ yang tidak nol", chk, C['green'] if ok else C['red'])]
    if tip:
        items.append(card("Apa yang terjadi selanjutnya?", [
            ("Titik putar", f"tumpuan {'B' if NA < 0 else 'A'} (x = {tip['pivot']:.2f} m)"), ("Στ terhadap titik putar", f"{tip['tau']:+.2f} N·m ({sense(tip['tau'])})"),
            ("I terhadap titik putar", f"{tip['I']:.3f} kg·m²"), ("α awal = Στ / I", f"{tip['alpha0']:+.2f} rad/s²  (ketidaksetimbangan → dinamika rotasi)")], C['red']))
    cards(*items)

EQ_DEFAULT = dict(L=2.0, Mb=4.0, xa=0.3, xb=1.7, m1=5.0, x1=0.7, m2=5.0, x2=1.3)
EQ_PRESETS = {
    "Simetris: N_A = N_B": {},
    "Beban 1 bergeser mendekati tumpuan B (N_B naik)": dict(m1=8.0, x1=1.6, m2=2.0, x2=1.0),
    "Mendekati batas guling (N_A ≈ 0)": dict(m1=9.0, x1=2.0, m2=0.0, x2=1.0),
    "Terguling: N_A < 0": dict(m1=12.0, x1=2.0, m2=0.0, x2=1.0),
}

def build_equilibrium():
    c = dict(L=S('Panjang balok L (m)', 2.0, 1.0, 3.0, 0.1, '.1f'), Mb=S('Massa balok (kg)', 4.0, 1, 10, 0.5, '.1f'),
             xa=S('Posisi tumpuan A, x_A (m)', 0.3, 0, 3, 0.05), xb=S('Posisi tumpuan B, x_B (m)', 1.7, 0, 3, 0.05),
             m1=S('Massa beban 1 (kg)', 5.0, 0, 20, 0.5, '.1f'), x1=S('Posisi beban 1, x₁ (m)', 0.7, 0, 3, 0.05),
             m2=S('Massa beban 2 (kg)', 5.0, 0, 20, 0.5, '.1f'), x2=S('Posisi beban 2, x₂ (m)', 1.3, 0, 3, 0.05))
    panel = Panel(c, draw_equilibrium)
    dd = w.Dropdown(options=list(EQ_PRESETS), description='Skenario cepat:', style={'description_width': '120px'}, layout=w.Layout(width='560px'))
    def on_preset(ch):
        vals = dict(EQ_DEFAULT); vals.update(EQ_PRESETS[ch['new']]); panel.set(**vals)
    dd.observe(on_preset, names='value')
    pred = w.FloatText(value=0.0, description='Prediksiku N_B (N):', style={'description_width': '140px'}, layout=w.Layout(width='300px'))
    pbtn = w.Button(description='Cek prediksi', button_style='warning', layout=w.Layout(width='130px')); pout = w.HTML()
    def check(_):
        v = panel.values(); L = v['L']; xa, xb, x1, x2 = [float(np.clip(v[k], 0, L)) for k in ('xa', 'xb', 'x1', 'x2')]
        if xb - xa < 0.05: return
        NA, NB = beam_reactions(L, v['Mb'], xa, xb, [(v['m1'], x1), (v['m2'], x2)])
        err = abs(pred.value - NB) / abs(NB) * 100 if abs(NB) > 1e-9 else float('nan')
        pout.value = (f"<div style='font-family:sans-serif;color:#3B2417;background:#FFF8EC;padding:8px 12px;border-radius:6px'>"
                      f"N_B sebenarnya = <b>{NB:.2f} N</b> (prediksimu {pred.value:.2f} N, selisih {err:.1f} %). "
                      f"Petunjuk: ambil poros di A → N_B·(x_B−x_A) = Σ (berat × lengan terhadap A).</div>")
    pbtn.on_click(check)
    info = info_box("📘 Konsep, model & asumsi (klik untuk membuka)", """
<b>Syarat kesetimbangan statis benda tegar:</b> ΣFx = 0, ΣFy = 0, dan Στ = 0 (terhadap sembarang titik acuan).<br>
Balok homogen bermassa M ditumpu di dua titik. Dari ΣFy = 0: N_A + N_B = W total. Dari Στ = 0 terhadap A: <b>N_B (x_B − x_A) = Σ wᵢ (xᵢ − x_A)</b>, dengan berat balok bekerja di titik tengahnya (L/2).
Panel verifikasi menghitung ulang Στ terhadap titik lain untuk menunjukkan bahwa hasilnya nol di mana pun.<br>
<b>Batas guling:</b> tumpuan hanya dapat <i>mendorong</i> (N ≥ 0). Bila perhitungan memberi N &lt; 0, tumpuan itu lepas dan balok mulai berotasi; di sini ditampilkan Στ dan α awal = Στ/I terhadap tumpuan lain — jembatan menuju dinamika rotasi.<br>
<b>Konteks (ilustratif):</b> balok dengan dua tumpuan dan beban berbentuk gerabah adalah situasi umum penyangga benda (papan/rak yang menopang barang). Ini <i>bukan</i> klaim tentang peralatan khusus pengrajin Kasongan.
Idealisasi: balok kaku homogen, beban titik, gaya horizontal diabaikan, tumpuan tanpa gesekan.""")
    guide = lab_guide('''
<b>Langkah:</b>
<ol><li><b>Prediksi</b>: pilih skenario <i>Beban 1 bergeser mendekati tumpuan B</i>. Tebak N_B (ambil poros di A: N_B·(x_B − x_A) = Σ berat × lengan terhadap A), isi kolom <i>Prediksiku N_B</i>, lalu klik <b>Cek prediksi</b>.</li>
<li><b>Amati</b>: panah merah = berat, panah biru kehijauan = gaya tumpuan. Kartu <i>Verifikasi</i> menunjukkan Στ terhadap A, B, dan pusat balok (≈ 0 berarti setimbang). Grafik bawah menunjukkan N_A dan N_B bila beban 1 digeser; daerah hijau = daerah setimbang.</li>
<li><b>Jelaskan</b>: mengapa gaya tumpuan terdekat dengan beban lebih besar?</li></ol>
<b>Tantangan:</b>
<ol><li>Skenario <i>Simetris</i>: N_A = N_B = 68,67 N. Geser beban 1 ke kanan; mana yang naik, N_A atau N_B?</li>
<li>Skenario <i>Beban 1 bergeser…</i>: hasilnya N_A ≈ 35,04 N dan N_B ≈ 102,30 N. Cocokkah dengan prediksimu?</li>
<li>Skenario <i>Mendekati batas guling</i>: N_A ≈ 0,70 N. Naikkan massa beban 1 dari 9,0 ke 9,5 kg: apa yang terjadi? Batas teorinya m = M(x_B − L/2)/(x₁ − x_B) = 4×0,7/0,3 ≈ 9,33 kg.</li>
<li>Skenario <i>Terguling</i> (m₁ = 12 kg): balok berotasi terhadap tumpuan B dengan α awal ≈ −1,79 rad/s² (searah jarum jam). Dari mana angka itu? (Στ/I pada kartu merah)</li></ol>
<b>Membaca hasil:</b> bila N &lt; 0, tumpuan tidak bisa menarik sehingga lepas dan balok tidak setimbang. Posisi di luar balok otomatis dibatasi pada ujung balok.''')
    ui = w.VBox([header("⑤ Kesetimbangan Benda Tegar: ΣF = 0 dan Στ = 0"),
                 w.HTML("<div style='font-family:sans-serif;font-size:13.5px;color:#3B2417'>🎯 <b>Tantangan:</b> geser beban 1 ke kanan. Pada posisi berapa tumpuan A tepat mulai lepas "
                        "(N_A = 0)? Perhatikan grafik bawah, lalu uji dengan skenario cepat.</div>"),
                 dd, w.HBox([w.VBox([c['L'], c['Mb'], c['xa'], c['xb']]), w.VBox([c['m1'], c['x1'], c['m2'], c['x2']])]),
                 w.HBox([pred, pbtn]), pout, panel.out, guide, info])
    panel.refresh()
    return ui, panel
```
